## Read Risk Log


```python
import pandas as pd, json
rows=[]
with open('outputs/risk_log.jsonl', 'r', encoding='utf-8') as f:
    for line in f:
        rows.append(json.loads(line))
df = pd.DataFrame(rows)
df.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,4))
plt.plot(df['timestamp_sec'], df['risk_score'], marker='o')
plt.xlabel('Time (seconds)')
plt.ylabel('Risk Score')
plt.title('CrowdGuard-RAG Risk Timeline')
plt.grid(True)
plt.show()
```
